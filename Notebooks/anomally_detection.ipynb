```python
# Get only normal traffic
normal_indices = np.where(y_label.values == "BENIGN")[0]

# Select 30000 normal samples
np.random.seed(42)
autoencoder_indices = np.random.choice(
    normal_indices,
    30000,
    replace=False
)

X_autoencoder = X_scaled[autoencoder_indices]

print("Autoencoder training data:", X_autoencoder.shape)
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

autoencoder = Sequential([
    Dense(32, activation="relu", input_shape=(78,)),
    Dense(16, activation="relu"),
    Dense(32, activation="relu"),
    Dense(78, activation="linear")
])

autoencoder.compile(
    optimizer="adam",
    loss="mse"
)

autoencoder.summary()
```

```python
history = autoencoder.fit(
    X_autoencoder,
    X_autoencoder,
    epochs=20,
    batch_size=256,
    validation_split=0.2,
    shuffle=True,
    verbose=1
)
```

```python
autoencoder.fit(X_autoencoder, X_autoencoder)
```

```python
X_reconstructed = autoencoder.predict(X_sample, verbose=0)

reconstruction_error = np.mean(
    np.square(X_sample - X_reconstructed),
    axis=1
)

print("Reconstruction errors:", reconstruction_error.shape)
print("Minimum error:", reconstruction_error.min())
print("Maximum error:", reconstruction_error.max())
print("Mean error:", reconstruction_error.mean())
```

```python
# Reconstruction error on normal training data
X_normal_reconstructed = autoencoder.predict(
    X_autoencoder,
    verbose=0
)

normal_reconstruction_error = np.mean(
    np.square(X_autoencoder - X_normal_reconstructed),
    axis=1
)

# Set threshold at the 95th percentile
threshold = np.percentile(
    normal_reconstruction_error,
    95
)

print("95% threshold:", threshold)
```

```python
# Predict Normal / Anomaly
autoencoder_pred = np.where(
    reconstruction_error > threshold,
    "Attack",
    "Normal"
)

print(pd.Series(autoencoder_pred).value_counts())
```
